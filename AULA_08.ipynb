Missão 02: Automação de Arquivos

Criar um programa que organize arquivos bagunçados em pastas automaticamente, basendo-se apenas na extensão do arquivo (ex: .pdf, .jpg).

```python
import os
import shutil

# 1. Defina o caminho da pasta que você quer organizar.
# O ponto '.' significa "a pasta onde esta este script está salvo".
diretorio = '.'

# 2. Lemos tudo o que existe dentro desse diretório
for arquivo in os.listdir(diretorio):

  # Criamos o caminho completo para o item atual
  caminho_completo = os.path.join(diretorio, arquivo)

  # 3. Verificamos se é um arquivo de verdade (para ignorar pastas que já existem)
  if os.path.isfile(caminho_completo):

    # 4. Dividimos o nome do arquivo da sua extensão (ex: 'foto', '.jpg')
    nome, extensao = os.path.splitext(arquivo)

    # Ignora arquivos sem extensão ou o próprio script Python (para ele não se mover)
    if not extensao or extensao == '.py':
      continue

    # 5. Tiramos o ponto da extensão para usar como nome da pasta (ex: de '.jpg' vira 'jpg')
    nome_pasta = extensao[1:].lower()
    caminho_pasta =  os.path.join(diretorio, nome_pasta)

    # 6. Se a pasta para essa extensão ainda não existir, o script cria
    if not os.path.exists(caminho_pasta):
        os.makedirs(caminho_pasta)

    # 7. Finalmente, movemos o arquivo para dentro da pasta correta
    caminho_destino = os.path.join(caminho_pasta, arquivo)
    shutil.move(caminho_completo, caminho_destino)

    print(f"Organizado: {arquivo} guardado na pasta '{nome_pasta}'")

print("Organizado concluída!")
```

Missão 03: Consulta Simples a APIs

Desenvolver um script que consulte uma API pública (ViaCEP) e retorne as informações de endereço formatadas de maneira legível.  

```python
import requests

def buscar_endereco(cep):
    """Consulta a API do ViaCEP e exibe o endereço formatado."""

    # 1. Limpeza básica
    cep = cep.replace("-", "").strip()

    # 2. Validação simples
    if len(cep) != 8 or not cep.isdigit():
        print("erro: O CEP informado é inválido. Digite 8 números.")
        return

    # 3. Montamos a URL
    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        # 4. Fazemos a requisição
        resposta = requests.get(url)
        resposta.raise_for_status()

        # 5. Transformamos a resposta
        dados = resposta.json()

        # 6. Checa se o CEP não existe
        if "erro" in dados:
            print(f"erro: O CEP {cep} não foi encontrado na base de dados.")
            return

        # 7. Exibimos os dados formatados
        print("\n === Informações do Endereço ===")
        # CORREÇÃO 1: Trocado 'bairro' por 'cep'
        print(f"CEP:        {dados.get('cep')}")
        print(f"Logradouro: {dados.get('logradouro')}")

        if dados.get('complemento'):
            print(f"Complemento: {dados.get('complemento')}")

        print(f"Bairro:    {dados.get('bairro')}")
        print(f"Cidade/UF: {dados.get('localidade')} - {dados.get('uf')}")
        print("================================\n")

    # 2 e 3: Removido as chaves de {e} e ajustado o alinhamento
    except requests.exceptions.RequestException as e:
        # Captura problemas como falta de internet
        print(f" Erro de conexão com a API: {e}")

# --- Testando o nosso script ---
cep_para_consultar = "01001-000"

# 4: Adicionado o comando que efetivamente chama a função
buscar_endereco(cep_para_consultar)
```

Missão 04: Sistema de Notificações

Criar um programa autônomo que dispare alertas (mensagens de alerta no console ou e-mail simulado) unicamente quando uma condição crítica for atingida.

```python
import time
import random

# --- Configurações do Sistema ---
LIMITE_CRITICO = 90.0  # O alerta só dispara se passar desse valor
INTERVALO_CHECAGEM = 2 # Segundos entre cada verificação

def simular_sensor():
    """Gera um número aleatório entre 50 e 100 para simular a leitura do sistema."""
    return random.uniform(50, 100)

def disparar_email_simulado(valor):
    """Simula o envio de um e-mail de alerta no console."""
    print("\n" + "="*50)
    print(" SIMULAÇÃO DE E-MAIL ENVIADO")
    print("Para: equipe.ti@empresa.com")
    print("Assunto:  URGENTE: Condição Crítica Detectada!")
    print("-" * 50)
    print(f"Mensagem: O sistema ultrapassou o limite seguro.")
    print(f"Valor registrado: {valor:.2f}% (Limite: {LIMITE_CRITICO}%)")
    print("Ação imediata é necessária!")
    print("="*50 + "\n")

def iniciar_monitoramento():
    """Função principal que roda em loop de forma autônoma."""
    print(f" Iniciando monitoramento autônomo...")
    print(f" Alertas configurados para valores acima de {LIMITE_CRITICO}%.")
    print("Pressione 'Ctrl + C' no terminal para parar o programa.\n")

    # O bloco try/except aqui serve para pararmos o programa elegantemente com Ctrl+C
    try:
        while True: # Este é o coração do programa autônomo (Loop Infinito)

            # 1. Fazemos a leitura daquele exato momento
            leitura_atual = simular_sensor()
            hora_atual = time.strftime('%H:%M:%S')

            # 2. Exibimos o status normal para sabermos que o script está rodando
            print(f"[{hora_atual}] Status do sistema: {leitura_atual:.2f}%")

            # 3. A regra de negócio: O alerta ÚNICO baseado na condição crítica
            if leitura_atual >= LIMITE_CRITICO:
                print(f"[{hora_atual}]  ALERTA: Pico detectado! Iniciando protocolos...")
                disparar_email_simulado(leitura_atual)

            # 4. Faz o programa pausar antes de checar de novo (evita sobrecarregar a máquina)
            time.sleep(INTERVALO_CHECAGEM)

    except KeyboardInterrupt:
        # Quando você aperta Ctrl+C, o loop quebra e cai aqui
        print("\n⏹️ Monitoramento interrompido pelo usuário. Sistema desligado.")

# --- Dando a partida no programa ---
iniciar_monitoramento()
```
